```python
# =============================================================================
# 3. EXPLORE DATA
# =============================================================================

# Load data - UPDATE THIS PATH
filepath = "../scripts/generated/data/nhl_data.csv"
df = pd.read_csv(filepath)

print(f"\nDataset shape: {df.shape}")
print(f"Columns in dataset: {len(df.columns)}")
print(f"Date range: {df['date'].min()} to {df['date'].max()}")

# Check target distribution
print("\n" + "="*60)
print("TARGET VARIABLE DISTRIBUTION")
print("="*60)
print(df['home_win'].value_counts())
print(f"\nHome win percentage: {df['home_win'].mean():.2%}")
```

```python
# =============================================================================
# 4. PREPARE FEATURES
# =============================================================================

# Check which features are available
available_features = [col for col in FEATURE_COLS if col in df.columns]
missing_features = [col for col in FEATURE_COLS if col not in df.columns]

print("\n" + "="*60)
print("FEATURE AVAILABILITY")
print("="*60)
print(f"Available features: {len(available_features)}/{len(FEATURE_COLS)}")

if missing_features:
    print(f"\nMissing features ({len(missing_features)}):")
    for feat in missing_features[:10]:
        print(f"  - {feat}")
    if len(missing_features) > 10:
        print(f"  ... and {len(missing_features) - 10} more")

# Prepare features and target
X = df[available_features].copy()
y = df['home_win'].copy()

# Keep date for splitting
dates = pd.to_datetime(df['date'])

print(f"\nFeatures shape: {X.shape}")
print(f"Target shape: {y.shape}")

# Check missing values
missing_counts = X.isnull().sum()
if missing_counts.sum() > 0:
    print(f"\nMissing values per feature:")
    print(missing_counts[missing_counts > 0])

# Handle missing values
print(f"\nMissing values before imputation: {X.isnull().sum().sum()}")

# X = X.dropna()

# # Fill with median
# X = X.fillna(X.median())

# # Handle any infinite values
# X = X.replace([np.inf, -np.inf], np.nan)
# X = X.fillna(X.median())

print(f"Missing values after imputation: {X.isnull().sum().sum()}")
```

```python
# shape
print(f"\nFinal features shape: {X.shape}")
print(f"Final target shape: {y.shape}")
```

```python
# =============================================================================
# 5. TEMPORAL TRAIN-TEST SPLIT
# Sports data is time-ordered — a random split leaks future games into training.
# We train on earlier seasons and test on the most recent season.
# =============================================================================

print("\n" + "="*60)
print("TEMPORAL TRAIN-TEST SPLIT")
print("="*60)

non_missing_idx = X.dropna().index
X_clean = X.loc[non_missing_idx].copy()
y_clean = y.loc[non_missing_idx].copy()
dates_clean = dates.loc[non_missing_idx].copy()

# Use the last full season (≥ Oct 2024) as the held-out test set
cutoff = pd.Timestamp("2024-10-01")
train_mask = dates_clean < cutoff
test_mask  = dates_clean >= cutoff

X_train, X_test = X_clean[train_mask], X_clean[test_mask]
y_train, y_test = y_clean[train_mask], y_clean[test_mask]

print(f"Train: {len(X_train)} games  ({dates_clean[train_mask].min().date()} → {dates_clean[train_mask].max().date()})")
print(f"Test:  {len(X_test)} games  ({dates_clean[test_mask].min().date()} → {dates_clean[test_mask].max().date()})")
print(f"Home win rate — train: {y_train.mean():.2%}  test: {y_test.mean():.2%}")
```

```python
# =============================================================================
# 7. HYPERPARAMETER TUNING — Random Forest with TimeSeriesSplit CV
# TimeSeriesSplit respects temporal order within the training window.
# =============================================================================

print("\n" + "="*60)
print("HYPERPARAMETER TUNING  (Random Forest)")
print("="*60)

tscv = TimeSeriesSplit(n_splits=5)

param_grid = {
    'n_estimators':      [100, 200, 300],
    'max_depth':         [10, 20, None],
    'min_samples_split': [2, 5],
    'min_samples_leaf':  [1, 2],
}

print(f"Total combinations: "
      f"{len(param_grid['n_estimators']) * len(param_grid['max_depth']) * len(param_grid['min_samples_split']) * len(param_grid['min_samples_leaf'])}")

rf = RandomForestClassifier(random_state=42, n_jobs=-1)

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=tscv,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1,
)

grid_search.fit(X_train, y_train)

print("\n" + "="*60)
print("BEST HYPERPARAMETERS")
print("="*60)
for param, value in grid_search.best_params_.items():
    print(f"  {param}: {value}")
print(f"\nBest CV Accuracy: {grid_search.best_score_:.4f}")

best_model = grid_search.best_estimator_
best_model.fit(X_train, y_train)


# =============================================================================
# 7B. XGBOOST COMPARISON
# =============================================================================

from xgboost import XGBClassifier

print("\n" + "="*60)
print("XGBOOST COMPARISON")
print("="*60)

xgb_param_grid = {
    'n_estimators':  [100, 300],
    'max_depth':     [4, 6],
    'learning_rate': [0.05, 0.1],
    'subsample':     [0.8, 1.0],
}

xgb = XGBClassifier(
    random_state=42,
    eval_metric='logloss',
    use_label_encoder=False,
    n_jobs=-1,
)

xgb_search = GridSearchCV(
    estimator=xgb,
    param_grid=xgb_param_grid,
    cv=tscv,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1,
)

xgb_search.fit(X_train, y_train)
best_xgb = xgb_search.best_estimator_
best_xgb.fit(X_train, y_train)

xgb_test_acc = accuracy_score(y_test, best_xgb.predict(X_test))
print(f"\nXGBoost best CV accuracy: {xgb_search.best_score_:.4f}")
print(f"XGBoost test accuracy:    {xgb_test_acc:.4f}")
for param, value in xgb_search.best_params_.items():
    print(f"  {param}: {value}")
```

```python
# =============================================================================
# 8. MODEL COMPARISON ON TEST SET
# =============================================================================

print("\n" + "="*60)
print("TEST SET PERFORMANCE")
print("="*60)

rf_acc  = accuracy_score(y_test, best_model.predict(X_test))
xgb_acc = accuracy_score(y_test, best_xgb.predict(X_test))

print(f"\nRandom Forest  — Accuracy: {rf_acc:.4f}")
print(classification_report(y_test, best_model.predict(X_test),
                             target_names=['Away Win', 'Home Win']))

print(f"XGBoost        — Accuracy: {xgb_acc:.4f}")
print(classification_report(y_test, best_xgb.predict(X_test),
                             target_names=['Away Win', 'Home Win']))

# Pick the better model for saving
if xgb_acc >= rf_acc:
    final_model = best_xgb
    model_name  = "XGBoost"
    test_accuracy = xgb_acc
else:
    final_model = best_model
    model_name  = "Random Forest"
    test_accuracy = rf_acc

print(f"\nSelected model: {model_name}  (test accuracy: {test_accuracy:.4f})")
y_test_pred = final_model.predict(X_test)
```

```python
# =============================================================================
# FEATURE IMPORTANCE ANALYSIS
# =============================================================================

print("\n" + "="*60)
print("FEATURE IMPORTANCE ANALYSIS")
print("="*60)

# Build dataframe
fi = pd.DataFrame({
    "feature": available_features,
    "importance": best_model.feature_importances_
}).sort_values("importance", ascending=False).head(20)

# Plot
plt.figure(figsize=(10, 6))
plt.barh(fi["feature"][::-1], fi["importance"][::-1])
plt.title("Top 20 Feature Importances")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```

```python
# =============================================================================
# 9. VISUALIZATIONS
# =============================================================================

print("\n" + "="*60)
print("CREATING VISUALIZATIONS")
print("="*60)

cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Away Win', 'Home Win'],
    yticklabels=['Away Win', 'Home Win'],
    cbar_kws={'label': 'Count'}
)
plt.title('Confusion Matrix - Test Set', fontsize=14, fontweight='bold')
plt.ylabel('True Label', fontsize=12)
plt.xlabel('Predicted Label', fontsize=12)
plt.tight_layout()
plt.show()
plt.close()
```
